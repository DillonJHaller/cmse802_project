# Reflectance Values

This notebook is used to generate some bar graphs showing reflectance and index values for the different LTPCs

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
#Set this to wherever the repo is saved
os.chdir("")
os.getcwd()
```

```python
#Read in training and testing data
train_df = pd.read_csv("data\\Training_Data.csv")
test_df = pd.read_csv("data\\Testing_Data.csv")
full_df = pd.concat([train_df, test_df], ignore_index=True)
full_df.head()
```

```python
#Run through all columns in the combined dataframe
for col in full_df.columns:
    #Plot bar chart of average value per LTPC
    plt.figure(figsize=(10, 6))
    grouped_df = full_df.groupby('LTPC')[col].mean()
    grouped_df.plot(kind='bar')
    plt.title(f'Average {col} by LTPC')
    plt.xlabel('LTPC')
    plt.ylabel(f'Average {col} Value')
    #Save figure
    plt.savefig(f'results\\figures\\barcharts\\average_{col}_by_LTPC.png')
    plt.close()
```
